# Ames Housing and Central Iowa Tornadoes

Load the two working CSV files into pandas DataFrames and explore their time ranges and basic summaries. The tornado DataFrame covers Story and six surrounding counties and is filtered to years at or after the earliest home construction year. This notebook reads the source CSVs without changing them.

```python
from pathlib import Path
import pandas as pd
```

```python
# Find the project root whether Jupyter starts in the repository root or analysis/.
project_root = next(
    (candidate for candidate in (Path.cwd(), *Path.cwd().parents)
     if (candidate / "analysis" / "AmesHousing-working.csv").exists()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Could not find analysis/AmesHousing-working.csv from the current directory.")

housing_path = project_root / "analysis" / "AmesHousing-working.csv"
tornado_path = project_root / "analysis" / "Iowa-central-counties-tornadoes-working.csv"
print(f"Project folder: {project_root}")
print(f"Housing file: {housing_path.name}")
print(f"Tornado file: {tornado_path.name}")
```

```python
# Keep both source tables as pandas DataFrames.
homes_df = pd.read_csv(housing_path)
tornadoes_df = pd.read_csv(tornado_path)

print(f"Homes: {homes_df.shape[0]:,} rows — {homes_df.shape[1]} columns")
print(f"Tornado records before filtering: {tornadoes_df.shape[0]:,}")
```

```python
# Remove tornado records from before the first home's construction year.
homes_df["Year Built"] = pd.to_numeric(homes_df["Year Built"], errors="coerce")
tornadoes_df["Year"] = pd.to_numeric(tornadoes_df["Year"], errors="coerce")
first_home_year = int(homes_df["Year Built"].min())
earliest_tornado_year = int(tornadoes_df["Year"].min())
excluded_tornadoes = tornadoes_df.loc[tornadoes_df["Year"] < first_home_year].copy()
tornadoes_df = tornadoes_df.loc[tornadoes_df["Year"] >= first_home_year].copy()

print(f"Earliest home construction year: {first_home_year}")
print(f"Earliest tornado year in the source file: {earliest_tornado_year}")
print(f"Tornadoes excluded as too early: {len(excluded_tornadoes)}")
print(f"Tornado records retained: {len(tornadoes_df)}")
```

```python
# Check the columns and the time spans now represented in each DataFrame.
display(homes_df[["Year Built", "Yr Sold", "SalePrice", "Total Bsmt SF"]].describe())
display(tornadoes_df[["Year", "FatalitiesTotal", "InjuriesTotal", "PropertyDamageUSD", "CropDamageUSD"]].describe())

print("Tornado intensity values (including missing ratings):")
display(tornadoes_df["Intensity"].value_counts(dropna=False).sort_index())
```

```python
# Initial housing summary by whether a recorded basement is present.
homes_df["Total Bsmt SF"] = pd.to_numeric(homes_df["Total Bsmt SF"], errors="coerce")
homes_df["Has Basement"] = homes_df["Total Bsmt SF"].fillna(0).gt(0)
basement_price_summary = (
    homes_df.groupby("Has Basement")["SalePrice"]
    .agg(Homes="count", MeanSalePrice="mean", MedianSalePrice="median")
    .rename(index={False: "No recorded basement", True: "Basement present"})
)
display(basement_price_summary)
```

```python
# Annual county-level tornado counts and impacts provide a starting point for later comparisons.
annual_tornado_summary = (
    tornadoes_df.groupby(["Year", "County"])
    .agg(
        Tornadoes=("EventID", "count"),
        Fatalities=("FatalitiesTotal", "sum"),
        Injuries=("InjuriesTotal", "sum"),
        PropertyDamageUSD=("PropertyDamageUSD", "sum"),
        CropDamageUSD=("CropDamageUSD", "sum"),
    )
)
display(annual_tornado_summary)
```

## Interpretation note

A tornado recorded in one of these counties during a calendar year does not establish that an individual Ames home was exposed to that tornado. Treat county and annual summaries as regional context; track locations and sale timing would be needed for a more specific exposure comparison.
